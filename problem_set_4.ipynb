```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

### Problem 4a

```python
df = pd.read_csv('data/heart_failure_clinical_records_dataset.csv')
df.head()
```

```python
plt.figure(figsize=(5, 4))
sns.kdeplot(df['ejection_fraction'])
plt.title('KDE of Ejection Fraction')
plt.show()
```

## Problem 4b

```python
plt.figure(figsize=(5, 4))
for i in range(15):
    df_sample = df.sample(frac=1.0, replace=True)
    sns.kdeplot(df_sample['ejection_fraction'],color='gray')

# Original plot overlayed for reference
sns.kdeplot(df['ejection_fraction'],color='black')
```
